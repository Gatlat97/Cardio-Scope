```python
import pandas as pd 
columns=[
 "age","sex","cp","trestbps","chol","fbs",
    "restecg","thalach","exang","oldpeak",
    "slope","ca","thal","num"
]

df = pd.read_csv(
    "data/processed.cleveland.data",
    header=None,
    names=columns,
    na_values="?"
)
# Keep only selected features and target
df = df[["age", "cp", "thalach", "oldpeak", "ca", "num"]]

# Convert target to binary
df["target"] = (df["num"] > 0).astype(int)

# Remove old target column
df.drop("num", axis=1, inplace=True)

# Fill missing values in ca
df["ca"] = df["ca"].fillna(df["ca"].median())

#convert features to integers
df["age"] =df["age"].astype(int)
df["cp"] = df["cp"].astype(int)
df["thalach"] = df["thalach"].astype(int)
df["ca"] = df["ca"].round().astype(int)

# Save cleaned data
df.to_csv("heart_disease_clean.csv", index=False)
```
